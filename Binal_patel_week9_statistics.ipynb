```python
sample_means = []

for i in range(100):
    sample = df['Sales'].sample(n=30)
    sample_means.append(sample.mean())
```

```python
plt.hist(sample_means, bins=20)
plt.xlabel("Sample Mean of Sales")
plt.ylabel("Frequency")
plt.title("Central Limit Theorem – Distribution of Sample Means")
plt.show()
```

```python
### Central Limit Theorem (CLT)

According to the Central Limit Theorem, the distribution of sample means
approaches a normal distribution as the sample size increases, regardless
of the shape of the original population distribution.

In this analysis, multiple samples of size 30 were taken from the Sales data.
The histogram of the sample means shows an approximately normal distribution,
which confirms the Central Limit Theorem.
```

```python
mean = df['Sales'].mean()
std_dev = df['Sales'].std()

mean, std_dev
```

```python
plt.hist(df['Sales'], bins=30)
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Sales Distribution")
plt.show()
```

```python
within_1_std = df[(df['Sales'] >= mean - std_dev) & (df['Sales'] <= mean + std_dev)]
within_2_std = df[(df['Sales'] >= mean - 2*std_dev) & (df['Sales'] <= mean + 2*std_dev)]
within_3_std = df[(df['Sales'] >= mean - 3*std_dev) & (df['Sales'] <= mean + 3*std_dev)]

len(within_1_std)/len(df), len(within_2_std)/len(df), len(within_3_std)/len(df)
```

```python
### Normal Distribution

The Sales data approximately follows a normal distribution, although it
shows slight skewness, which is common in real-world business data.

Using the empirical rule:
- About 68% of the data lies within one standard deviation of the mean
- About 95% lies within two standard deviations
- About 99% lies within three standard deviations

This confirms that the Sales data approximately follows a normal distribution.
```

```python
df['Z_Score'] = (df['Sales'] - mean) / std_dev
df[['Sales', 'Z_Score']].head()
```

```python
outliers = df[(df['Z_Score'] > 3) | (df['Z_Score'] < -3)]
outliers[['Sales', 'Z_Score']].head()
```

```python
len(outliers)
```

```python
### Z-Score Analysis

Z-score measures how many standard deviations a data point is from the mean.

In this analysis, Z-scores were calculated for the Sales data. Any value with
a Z-score greater than +3 or less than –3 was treated as an outlier.

These outliers represent unusually high or low Sales values and may require
further investigation in a real business scenario.
```
